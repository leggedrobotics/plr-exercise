# Install and Import Required Libraries
Install wandb using pip and import the necessary libraries.

```python
# Import the necessary ibraries
import wandb
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os
```

# Initialize wandb Project
Initialize a new wandb project and configure the settings.

```python
# Initialize wandb Project
wandb.init(project="plr-exercise", config={
    "learning_rate": 0.001,
    "epochs": 10,
    "batch_size": 32
})

# Access the configuration
config = wandb.config

# Log the configuration
wandb.config.update({
    "optimizer": "adam",
    "loss_function": "categorical_crossentropy"
})
```

# Log Random Curves
Generate random curves and log them to wandb.

```python
# Generate random curves and log them to wandb
x = np.linspace(0, 10, 100)
y1 = np.sin(x) + np.random.normal(0, 0.1, 100)
y2 = np.cos(x) + np.random.normal(0, 0.1, 100)

plt.figure()
plt.plot(x, y1, label='Random Sine Curve')
plt.plot(x, y2, label='Random Cosine Curve')

plt.legend()
plt.title('Random Curves')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

# Save the plot as an image file
plt.savefig('random_curves.png')

# Log the plot image to wandb
wandb.log({"random_curves": wandb.Image('random_curves.png')})

# Optionally, log the data points as well
wandb.log({"x": x, "y1": y1, "y2": y2})
```
